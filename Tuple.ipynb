1.A list contains names of boys and girls as its elements. Boys’ names are stored as tuples. Write a
program to find out number of boys and girls in the list. (Hint: use isinstance(ele,tuple))

```python
names_list = [("John", "Doe"), ("Alice",), "Sarah", ("Mike", "Smith"), "Emma", "David", ("Tom", "Hanks")]

# Initialize counters for boys and girls
boys_count = 0
girls_count = 0

# Loop through the list and count boys and girls
for ele in names_list:
    if isinstance(ele, tuple):  # If the element is a tuple, it's a boy
        boys_count += 1
    else:  # If it's not a tuple, it's a girl
        girls_count += 1

# Output the counts
print("Number of boys:", boys_count)
print("Number of girls:", girls_count)
```

2.A list contains tuples containing roll no., name and age of student. Write a python program to create
three lists separately for roll no., name and age

```python
n = int(input("Enter number of students: "))

students = []
roll_nos = []
names = []
ages = []

for i in range(n):
    print("Enter details for student", i + 1)
    roll = int(input("Roll no: "))
    name = input("Name: ")
    age = int(input("Age: "))
    students.append((roll, name, age))

for student in students:
    roll_nos.append(student[0])
    names.append(student[1])
    ages.append(student[2])


print("Roll numbers:", roll_nos)
print("Names:", names)
print("Ages:", ages)
```

3.Suppose a date is represented as a tuple (d, m, y). Create two date tuples and find the number of
days between the two dates.

```python
d1 = int(input("Enter day of first date: "))
m1 = int(input("Enter month of first date: "))
y1 = int(input("Enter year of first date: "))

d2 = int(input("Enter day of second date: "))
m2 = int(input("Enter month of second date: "))
y2 = int(input("Enter year of second date: "))

def to_days(d, m, y):
    return y * 365 + m * 30 + d

total1 = to_days(d1, m1, y1)
total2 = to_days(d2, m2, y2)
2
diff = abs(total2 - total1)

print("Approximate number of days between the two dates:", diff)
```

4.Create a list of tuples containing a food item and its price. Sort the tuples in descending order by
price.

```python
food_items = [("Burger", 150), ("Pizza", 250), ("Fries", 100), ("Sandwich", 120), ("Pasta", 180)]

sorted_items = sorted(food_items, key=lambda x: x[1], reverse=True)

for item in sorted_items:
    print(item[0], "-", item[1])
```

5.Remove empty tuple(s) from the list of tuples.

```python
tuple_list = [(), (1, 2), (), (3,), (4, 5, 6), (), ("a",), ()]

cleaned_list = [t for t in tuple_list if t != ()]

print("List after removing empty tuples:", cleaned_list)
```

6.Modify an element of a tuple.

```python
original_tuple = (10, 20, 30, 40)

temp_list = list(original_tuple)

temp_list[1] = 99


modified_tuple = tuple(temp_list)

print("Original tuple:", original_tuple)
print("Modified tuple:", modified_tuple)
```
