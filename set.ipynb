1.Write a program that converts words present in a list into uppercase and stores them in a set.

```python
words = input("Enter words separated by spaces: ").split()

upper_set = set()
for word in words:
    upper_set.add(word.upper())


print("Uppercase words in a set:", upper_set)
```

2.Write a program to create a set containing 10 random numbers in the range 15 to 45. Count how
many of these numbers are less than 30. Delete all numbers that are greater than 35.

```python
input_str = input("Enter 10 numbers between 15 and 45 separated by spaces: ")
numbers = [int(x) for x in input_str.split()]

if len(numbers) != 10:
    print("Please enter exactly 10 numbers.")
else:

    num_set = set(numbers)
    print("Original set:", num_set)


    count_less_than_30 = 0
    for num in num_set:
        if num < 30:
            count_less_than_30 += 1
    print("Count of numbers less than 30:", count_less_than_30)


    filtered_set = set()
    for num in num_set:
        if num <= 35:
            filtered_set.add(num)
    print("Set after removing numbers greater than 35:", filtered_set)
```

3.Create an empty set. Write a program that adds five new names to this set, modifies one existing
name and deletes two names from it.

```python
name_set = set()

for i in range(5):
    name = input("Enter name " + str(i + 1) + ": ")
    name_set.add(name)

print("Set after adding names:", name_set)


old_name = input("Enter the name you want to modify: ")
if old_name in name_set:
    name_set.remove(old_name)
    new_name = input("Enter the new name to replace '" + old_name + "': ")
    name_set.add(new_name)
else:
    print("Name '" + old_name + "' not found in the set.")

print("Set after modifying a name:", name_set)


for i in range(2):
    name_to_delete = input("Enter the name you want to delete (#" + str(i + 1) + "): ")
    if name_to_delete in name_set:
        name_set.remove(name_to_delete)
    else:
        print("Name '" + name_to_delete + "' not found in the set.")

print("Set after deleting two names:", name_set)
```

4.A set contains names which begin either with A or with B. Write a program to separate out the
names into two sets, one containing names beginning with A and another with B.

```python
input_str = input("Enter names separated by spaces (starting with A or B only): ")
name_set = set(input_str.split())

a_names = set()
b_names = set()


for name in name_set:
    if name.startswith("A"):
        a_names.add(name)
    elif name.startswith("B"):
        b_names.add(name)

print("Names starting with A: " + str(a_names))
print("Names starting with B: " + str(b_names))
```
