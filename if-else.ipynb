10.Given the length and breadth of a rectangle, write a program to find whether the are of the rectangle is greater than its perimeter.

```python
length=float(input("enter length of rectangle"))
breadth=float(input("enter breadth of rectangle"))

area=length*breadth
perimeter=2*(length+breadth)

if area>perimeter:
  print("area is greater than perimeter")
else:
  print("area is not greater than perimeter")
```

12.Given the coordinates (x,y) of center of a circle and its radius, determine whether a point lies inside the circle, on the circle or outside the circle. (Hint: Use sqrt( ), pow( ) )

```python
h=int(input("enter x-coordinate of circle center"))
k=int(input("enter y-coordinate of circle center"))
r=int(input("enter radius of circle"))

x=int(input("enter x coordinate of point"))
y=int(input("enter y coordinate of point"))

distance_squared = (x - h) * (x - h) + (y - k) * (y - k)
radius_squared = r * r

if distance_squared < radius_squared:
    print("The point is inside the circle.")
elif distance_squared == radius_squared:
    print("The point is on the circle.")
else:
    print("The point is outside the circle.")
```

12.Convert number 0 to 19 to its equivalent words. E.g. 0 → zero, 19→nineteen.
