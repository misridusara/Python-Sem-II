3) Count no. of alphabets and no. of digits in any given string.

```python
str=input("enter a string")

alphabets=0
digits=0
for i in str:
    if i.isalpha():
        alphabets=alphabets+1
    elif i.isdigit():
        digits=digits+1
print('alphabets=',alphabets)
print('digits=',digits)
```

4) Check whether a given number is prime, is perfect, is Armstrong, is palindrome, is automorphic.

```python
num=int(input("enter a number"))

if num > 1:
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
else:
    is_prime = False

sum_divisors = 0
for i in range(1, num):
    if num % i == 0:
        sum_divisors += i
is_perfect = sum_divisors == num

temp = num
sum_armstrong = 0
digits = 0
while temp > 0:
    digits += 1
    temp //= 10

temp = num
while temp > 0:
    digit = temp % 10
    sum_armstrong += digit ** digits
    temp //= 10
is_armstrong = sum_armstrong == num

temp = num
reverse_num = 0
while temp > 0:
    reverse_num = reverse_num * 10 + temp % 10
    temp //= 10
is_palindrome = reverse_num == num

square = num * num
temp = num
match = True
while temp > 0:
    if temp % 10 != square % 10:
        match = False
        break
    temp //= 10
    square //= 10
is_automorphic = match

print("Prime:", "Yes" if is_prime else "No")
print("Perfect:", "Yes" if is_perfect else "No")
print("Armstrong:", "Yes" if is_armstrong else "No")
print("Palindrome:", "Yes" if is_palindrome else "No")
print("Automorphic:", "Yes" if is_automorphic else "No")
```

5) Generate all Pythagorean Triplets with side length <= 30.

```python
for a in range(1, 31):
    for b in range(a, 31):
      for c in range(b, 31):
        if a * a + b * b == c * c:
          print(f"({a}, {b}, {c})")
```

6) Print 24 hours of day with suitable suffixes like AM, PM, Noon and Midnight.

```python
for hour in range(24):
  if hour==0:
    print("12 Midnight")
  elif hour==12:
    print("12 Noon")
  elif hour<12:
    print(hour,"AM")
  else:
    print(hour-12,"PM")
```

```python
n=int(input("enter n:"))
r=int(input("enter r:"))

fact_n=1
for i in range(1,n+1):
  fact_n*=i

fact_r=1
for i in range(1,r+1):
  fact_r*=i

fact_n_r=1
for i in range(1,n-r+1):
  fact_n_r*=i

nPr = fact_n // fact_n_r
nCr = fact_n // (fact_r * fact_n_r)

print("nPr=",nPr)
print("nCr=",nCr)
```
