and print the list. Provide appropriate message at each stage.

6.Convert list of temperatures in Fahrenheit degrees to equivalent Celsius degrees.

```python
fahrenheit_temps = [32, 45, 68, 77, 104, 122]

print("Temperatures in Fahrenheit:")
print(fahrenheit_temps)


celsius_temps = []
for temp in fahrenheit_temps:
    celsius = (5 / 9) * (temp - 32)
    celsius_temps.append(round(celsius, 2))

print("\nEquivalent temperatures in Celsius:")
print(celsius_temps)
```

```python
stack=[]

while true:
  print("\nstack menu:")
  print("1.push")
  print("2.pop")
  print("3.peek")
  print("4.display")
  print("5.quit")

  choice=input("enter your choice(1-5):")

  if choice=="1":
    item=input("enter the element to push:")
    stack.append(item)
    print(item,"has been pushed to the stack.")

  elif choice=="2":
    if len(stack)==0:
      print("the stack is empty.")
    else:
      item=stack.pop()
      print(item,"has been popped from the stack.")

  elif choice=="3":
    if len(stack)==0:
      print("the stack is empty.")
    else:
      peek_item=stack[-1]
      print("the top element of the stack is:",peek_item)

  elif choice=="4":
    if not stack:
      print("stack is empty")
    else:
      print("cureent stack:")
      for item in reversed(stack):
        print(item)

  elif choice=="5":
    print("exiting program")
    break

  else:
    print("invalid choice")
```
